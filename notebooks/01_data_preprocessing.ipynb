# Phân loại bình luận độc hại tiếng Việt

## Thư viện

```python
# import thu vien
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

## Tải dữ liệu

```python
# lay thu muc goc cua project
PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists() and (PROJECT_ROOT.parent / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

# duong dan du lieu
BASE_DIR = str(PROJECT_ROOT / "data")
RAW_DIR = os.path.join(BASE_DIR, "raw")
PROCESSED_DIR = os.path.join(BASE_DIR, "processed")

TRAIN_PATH = os.path.join(RAW_DIR, "train_raw.csv")
VAL_PATH = os.path.join(RAW_DIR, "val_raw.csv")
TEST_PATH = os.path.join(RAW_DIR, "test_raw.csv")

TEXT_COL = "free_text"
LABEL_COL = "label_id"

LABEL_MAP = {
    0: "CLEAN",
    1: "OFFENSIVE",
    2: "HATE"
}
```

```python
# doc du lieu
train_df = pd.read_csv(TRAIN_PATH)
val_df = pd.read_csv(VAL_PATH)
test_df = pd.read_csv(TEST_PATH)

assert TEXT_COL in train_df.columns and LABEL_COL in train_df.columns
assert TEXT_COL in val_df.columns and LABEL_COL in val_df.columns
assert TEXT_COL in test_df.columns and LABEL_COL in test_df.columns

train_df.shape, val_df.shape, test_df.shape
```

## Kiểm tra dữ liệu

```python
# xem du lieu
train_df.head()
```

```python
# kiem tra null va phan bo nhan
print(train_df[[TEXT_COL, LABEL_COL]].isnull().sum())
print()
print(train_df[LABEL_COL].value_counts().sort_index())
```

```python
# check for raw overlaps between datasets
train_raw_texts = set(train_df[TEXT_COL].fillna("").astype(str).values)
val_raw_texts = set(val_df[TEXT_COL].fillna("").astype(str).values)
test_raw_texts = set(test_df[TEXT_COL].fillna("").astype(str).values)

train_val_overlap_raw = len(train_raw_texts & val_raw_texts)
train_test_overlap_raw = len(train_raw_texts & test_raw_texts)
val_test_overlap_raw = len(val_raw_texts & test_raw_texts)

print("=== RAW DATA OVERLAPS ===")
print(f"Train-Val overlap: {train_val_overlap_raw}")
print(f"Train-Test overlap: {train_test_overlap_raw}")
print(f"Val-Test overlap: {val_test_overlap_raw}")
print(f"\nTotal unique texts:")
print(f"Train: {len(train_raw_texts)}, Val: {len(val_raw_texts)}, Test: {len(test_raw_texts)}")
print(f"Combined unique: {len(train_raw_texts | val_raw_texts | test_raw_texts)}")
```

## Khám phá dữ liệu

```python
# ve phan bo nhan
label_count = train_df[LABEL_COL].value_counts().sort_index()
label_name = [LABEL_MAP.get(i, str(i)) for i in label_count.index]

plt.figure(figsize=(7, 4))
plt.bar(label_name, label_count.values)
plt.title("Train label distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.show()
```

```python
# thong ke do dai
train_text = train_df[TEXT_COL].fillna("").astype(str)

char_len = train_text.str.len()
word_len = train_text.str.split().str.len()

char_p99 = char_len.quantile(0.99)
word_p99 = word_len.quantile(0.99)

length_stats = pd.DataFrame({
    "stat": ["mean", "median", "p95", "p99", "max"],
    "char_len": [
        round(char_len.mean(), 2),
        round(char_len.median(), 2),
        round(char_len.quantile(0.95), 2),
        round(char_p99, 2),
        int(char_len.max())
    ],
    "word_len": [
        round(word_len.mean(), 2),
        round(word_len.median(), 2),
        round(word_len.quantile(0.95), 2),
        round(word_p99, 2),
        int(word_len.max())
    ]
})

length_stats
```

```python
# ve do dai binh luan
plt.figure(figsize=(7, 4))
plt.hist(char_len[char_len <= char_p99], bins=50)
plt.title("Train char length")
plt.xlabel("Chars")
plt.ylabel("Count")
plt.show()

plt.figure(figsize=(7, 4))
plt.hist(word_len[word_len <= word_p99], bins=50)
plt.title("Train word length")
plt.xlabel("Words")
plt.ylabel("Count")
plt.show()
```

## Tiền xử lý

```python
# ham xu ly text
url_pattern = re.compile(r"https?://\S+|www\.\S+")

def clean_text(text):
    if pd.isna(text):
        return np.nan
    text = str(text)
    text = text.replace("\n", " ").replace("\t", " ")
    text = url_pattern.sub("[URL]", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def to_label_id(value):
    if pd.isna(value):
        return np.nan
    if isinstance(value, str):
        value = value.strip().upper()
        if value in ["CLEAN", "OFFENSIVE", "HATE"]:
            return {"CLEAN": 0, "OFFENSIVE": 1, "HATE": 2}[value]
    try:
        value = int(value)
        if value in LABEL_MAP:
            return value
    except:
        pass
    return np.nan

def preprocess(df):
    data = df[[TEXT_COL, LABEL_COL]].copy()
    data["text_original"] = data[TEXT_COL]
    data["text_clean"] = data[TEXT_COL].apply(clean_text)
    data = data.dropna(subset=["text_clean", LABEL_COL]).copy()
    data = data[data["text_clean"].str.strip() != ""].copy()
    data["label_id"] = data[LABEL_COL].apply(to_label_id)
    data = data.dropna(subset=["label_id"]).copy()
    data["label_id"] = data["label_id"].astype(int)
    data["label_name"] = data["label_id"].map(LABEL_MAP)
    data["label_binary"] = data["label_id"].apply(lambda x: 0 if x == 0 else 1)
    return data[["text_original", "text_clean", "label_id", "label_name", "label_binary"]]

train_processed = preprocess(train_df)
val_processed = preprocess(val_df)
test_processed = preprocess(test_df)

train_processed.shape, val_processed.shape, test_processed.shape
```

```python
# xem du lieu sau xu ly
train_processed.head()
```

```python
# check for duplicates in each dataset
train_duplicates = train_processed['text_original'].duplicated().sum()
val_duplicates = val_processed['text_original'].duplicated().sum()
test_duplicates = test_processed['text_original'].duplicated().sum()

print(f"Train duplicates: {train_duplicates}")
print(f"Val duplicates: {val_duplicates}")
print(f"Test duplicates: {test_duplicates}")

# check for overlaps between datasets
train_texts = set(train_processed['text_clean'].values)
val_texts = set(val_processed['text_clean'].values)
test_texts = set(test_processed['text_clean'].values)

train_val_overlap = len(train_texts & val_texts)
train_test_overlap = len(train_texts & test_texts)
val_test_overlap = len(val_texts & test_texts)

print(f"\nTrain-Val overlap: {train_val_overlap}")
print(f"Train-Test overlap: {train_test_overlap}")
print(f"Val-Test overlap: {val_test_overlap}")
```

```python
# remove duplicates from processed datasets
train_processed_clean = train_processed.drop_duplicates(subset=['text_clean']).copy()
val_processed_clean = val_processed.drop_duplicates(subset=['text_clean']).copy()
test_processed_clean = test_processed.drop_duplicates(subset=['text_clean']).copy()

print("=== DUPLICATES REMOVED ===")
print(f"Train: {len(train_processed)} -> {len(train_processed_clean)} (removed {len(train_processed) - len(train_processed_clean)})")
print(f"Val: {len(val_processed)} -> {len(val_processed_clean)} (removed {len(val_processed) - len(val_processed_clean)})")
print(f"Test: {len(test_processed)} -> {len(test_processed_clean)} (removed {len(test_processed) - len(test_processed_clean)})")

# verify no duplicates
train_check = train_processed_clean['text_clean'].duplicated().sum()
val_check = val_processed_clean['text_clean'].duplicated().sum()
test_check = test_processed_clean['text_clean'].duplicated().sum()

print(f"\nVerification - Remaining duplicates:")
print(f"Train: {train_check}, Val: {val_check}, Test: {test_check}")
```

## Lưu dữ liệu

```python
# luu file processed (duplicates removed)
os.makedirs(PROCESSED_DIR, exist_ok=True)

train_processed_clean.to_csv(os.path.join(PROCESSED_DIR, "train_processed.csv"), index=False)
val_processed_clean.to_csv(os.path.join(PROCESSED_DIR, "val_processed.csv"), index=False)
test_processed_clean.to_csv(os.path.join(PROCESSED_DIR, "test_processed.csv"), index=False)

print(f"\nFiles saved:")
print(f"Train: {len(train_processed_clean)} samples")
print(f"Val: {len(val_processed_clean)} samples")
print(f"Test: {len(test_processed_clean)} samples")
```

## Class balance status post processing

```python
# class balance sau xu ly cho train/val/test
datasets = {
    "Train": train_processed_clean,
    "Val": val_processed_clean,
    "Test": test_processed_clean
}

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

for ax, (split_name, split_df) in zip(axes, datasets.items()):
    counts = split_df[LABEL_COL].value_counts().sort_index()
    labels = [LABEL_MAP.get(i, str(i)) for i in counts.index]
    
    ax.bar(labels, counts.values)
    ax.set_title(f"{split_name} label distribution")
    ax.set_xlabel("Label")
    ax.set_ylabel("Count")
    
    for i, v in enumerate(counts.values):
        ax.text(i, v, str(v), ha="center", va="bottom", fontsize=9)

plt.tight_layout()
plt.show()

for split_name, split_df in datasets.items():
    print(f"\n{split_name}:")
    print(split_df[LABEL_COL].value_counts().sort_index().rename(index=LABEL_MAP))
```
